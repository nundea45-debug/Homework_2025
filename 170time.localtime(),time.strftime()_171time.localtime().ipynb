```python
# Unit 170 현재 시간을 연-월-일 시:분:초로 출력하기 : time.localtime(), time.strftime()

# localtime() => 대한민국의 현재시간을 time.struct_time 자료형으로 리턴
```

```python
# 2024년 07월 02일 22시 05분 46초의 struct_time의 예
(tm_year=2024, tm_mon=7, tm_mday=2, tm_hour=22, tm_min=5, tm_sec=46, tm_wday=1, tm_yday=184, tm_isdst=0)
```

```python
# 현재 시간을 YYYY-MM-DD hh:mm:ss로 출력하는 예제
from time import localtime, strftime

time_stamp = strftime('%y-%m-%d %x', localtime())  # localtime()을 포맷문자열에 대응하는 값으로 할당하여 나온 문자열을 리턴
print(f'현재시간 : {time_stamp}')
```

```python
# Unit 171 올해의 경과된 날짜 수 계산 : time.localtime()

# struct_time에서 tm_yday 항목  => 당해년도 1월 1일부터 오늘까지 경과한 날짜수
```

```python
from time import localtime, strftime

t = localtime()  # t => localtime()이 리턴값을 struct_time 자료형으로 리턴받을 값
today = strftime('%y-%m-%d', t)  # 올해 1월 1일부터 오늘까지 경과한 일수가 저장된 값
print(f'{today}은 {t.tm_year}-01-01 이후 {t.tm_yday}일째 되는 날입니다.')
```
