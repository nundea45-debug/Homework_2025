```python
# 주어진 정수 n까지 누적합을 구해서 리턴하는 함수를 구현한 코드
def cumalativeSum(n) :
  ret = 0
  for i in range(1, n+1) :
    ret += i
  return ret
```

```python
from time import time

n = 10000000
stime = time()  # 누적합 계산 함수 실행 전 time() 값을 stime()으로 설정
answer = cumalativeSum(n)  # 1에서 10,000,000까지 누적합을 계산하는 함수 실행
etime = time()  # cumalativeSum() 종료 -> time() 값을 etime으로 둠
print(f'1에서 {n:,}까지 누적합 계산 시간 : {(etime-stime):.3f}초')  # cumalativeSum()가 수행된 소요 시간 => etime-stime이 됨(소수 셋째 자리까지 출력)
```

```python
# '*'을 출력하고 0.5초 동안 일시정지
from time import sleep

for i in range(1, 6):
  print('*'*i)
  sleep(0.5)
```
