```python
# Unit 172 오늘의 요일 계산하기 : time.localtime()

# tm_wday => 0~6가지 정수값을 가짐(0이 월요일, 6은 일요일을 뜻함)
```

```python
from time import localtime, strftime

weekdays = ['월요일', '화요일', '수요일', '목요일', '금요일', '토요일', '일요일']  # '월요일 ~ 일요일'까지 문자열을 요소로하는 리스트 정의
t = localtime()
today = strftime('%y-%m-%d', t)
print(f'{today}은 {weekdays[t.tm_wday]}입니다.')  # t,tm_wdy가 weekdays의 인덱스로 입력된 weekdays는 해당 요일 출력
```
